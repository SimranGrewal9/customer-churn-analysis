```python
# Import required libraries for data analysis and visualization 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\LENOVO\OneDrive\Desktop\WA_Fn-UseC_-Telco-Customer-Churn.csv")  #load the customer churn dataset from file path
print(df.head())  #display first 5 rows to understand dataset structure
```

```python
df.isnull().sum #check for missing values in each column
```

```python
df.info() #check dataset information like column types, nonnull values, memory usage
```

```python
#Visualize distribution of churn(target variable), showing how many customers left vs stayed
sns.countplot(x="Churn", data=df)
plt.title("Churn Distribution")
plt.show()
```

```python
#Compare monthly charges between churned and non-churned customers, helps understanding if higher charges affect churn
sns.boxplot(x="Churn", y="MonthlyCharges", data=df)
plt.title("Monthly Charges vs Churn")
plt.show()
```

```python
#Analyze contract type and its relation with churn, helps identify if shortterm contracts lead to higher churn
sns.countplot(x="Contract", hue="Churn", data=df)
plt.xticks(rotation=15)
plt.title("Contract Type vs Churn")
plt.show()
```

```python
#Plot age distribution or senior citizen distribution depending on citizen distribution depending on dataset, helps understand customer age pattern
sns.countplot(x="SeniorCitizen", data=df)
plt.title("Senior Citizen Distribution")
plt.show()
```

```python
# Analyze internet service type and its impact on churn, helps identify which service type has more churn
sns.countplot(x="InternetService", hue="Churn", data=df)
plt.xticks(rotation=15)
plt.title("Internet Service vs Churn")
plt.show()
```

```python
#Analyze payment methods and churn behaviour, helps identify if certain payment methods lead to higher churn
sns.countplot(x="PaymentMethod", hue="Churn", data=df)
plt.xticks(rotation=20)
plt.title("Payment Method vs Churn")
plt.show()
```

```python
# Show correlation between numerical features, helps understand relationships between variables 
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
#Final summary of dataset showing shape, missing values, churn percentage
print("Dataset Shape:", df.shape)
print("Total Missing Values", df.isnull().sum().sum())
print("Churn Percentage:")
print(df["Churn"].value_counts(normalize=True) * 100)
```
